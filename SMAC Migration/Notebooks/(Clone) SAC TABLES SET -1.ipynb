```python
%sql
CREATE OR REPLACE VIEW reporting_layer.sac_prod_seafarer_public.appraisal_performance AS

WITH base AS (
    SELECT
        *,
        from_json(feedback,'array<string>') AS feedback_array
    FROM landing_zone.db_sac_prod_seafarer_public.appraisals
    WHERE feedback IS NOT NULL  and status = 'Closed'
),

feedback_exploded AS (
    SELECT
        a.seafarer_id,
        a.contract_id,
        a.vessel_id,
        a.created_at,
        a.to_date,
        a.from_date,
        f.feedback_obj
    FROM base a
    LATERAL VIEW explode(feedback_array) f AS feedback_obj
),

feedback_parsed AS (
    SELECT
        seafarer_id,
        contract_id,
        vessel_id,
        created_at,
        to_date,
        from_date,
        get_json_object(feedback_obj,'$.appraisal_template_id') AS template_id,
        get_json_object(feedback_obj,'$.templateName') AS template_name,
        get_json_object(feedback_obj,'$.templateType') AS template_type,
        get_json_object(feedback_obj,'$.appraiser_name') AS appraiser_name,
        get_json_object(feedback_obj,'$.appraiser_email') AS appraiser_email,
        get_json_object(feedback_obj,'$.rating') AS rating,
        get_json_object(feedback_obj,'$.status') AS status,
        get_json_object(feedback_obj,'$.responded_at') AS responded_at,
        from_json(
            get_json_object(feedback_obj,'$.response'),
            'MAP<STRING,STRING>'
        ) AS response_map
    FROM feedback_exploded
),

question_level AS (
    SELECT
        *,
        q.key AS question_key,
        q.value AS question_value
    FROM feedback_parsed
    LATERAL VIEW OUTER inline(map_entries(response_map)) q
),

row_level AS (
    SELECT
        *,
        from_json(question_value,'MAP<STRING,STRING>') AS row_map
    FROM question_level
),

row_exploded AS (
    SELECT
        *,
        r.key AS row_key,
        r.value AS row_value
    FROM row_level
    LATERAL VIEW OUTER inline(map_entries(row_map)) r
),

column_level AS (
    SELECT
        *,
        from_json(row_value,'MAP<STRING,STRING>') AS column_map
    FROM row_exploded
),

matrix_output AS (
    SELECT
        seafarer_id,
        contract_id,
        created_at,
        vessel_id,
        to_date,
        from_date,
        template_id,
        template_name,
        template_type,
        appraiser_name,
        appraiser_email,
        rating,
        status,
        responded_at,
        question_key,
        row_key,
        c.key AS column_key,
        c.value AS response_value,
        NULL AS comment_value
    FROM column_level
    LATERAL VIEW OUTER inline(map_entries(column_map)) c
),

comment_output AS (
    SELECT
        seafarer_id,
        contract_id,
        created_at,
        vessel_id,
        to_date,
        from_date,
        template_id,
        template_name,
        template_type,
        appraiser_name,
        appraiser_email,
        rating,
        status,
        responded_at,
        question_key,
        NULL AS row_key,
        NULL AS column_key,
        NULL AS response_value,
        question_value AS comment_value
    FROM question_level
    WHERE question_value NOT LIKE '{%'
)

SELECT
    seafarer_id,
    created_at,
    crew_code,
    to_date,
    from_date,
    ROUND(AVG(performance_rating),2) AS performance_rating

FROM (
    SELECT
        ap.seafarer_id,
        ap.contract_id,
        ap.created_at,
        ap.vessel_id,
        ap.to_date,
        ap.from_date,
        ap.template_id,
        ap.template_name,
        ap.status,
        AVG(CAST(ap.rating AS DOUBLE)) AS performance_rating,
        sf.crew_code
    FROM (
        SELECT * FROM matrix_output
        UNION ALL
        SELECT * FROM comment_output
    ) ap
    LEFT JOIN landing_zone.db_sac_prod_seafarer_public.seafarers sf
        ON ap.seafarer_id = sf.id

    -- LEFT JOIN reporting_layer.sac_prod_seafarer_public.bi_inactive_seafarers I
    --       ON I.ID = sf.id
    -- WHERE I.ID is null -- this is to eliminate the inactive seafarers
    
    GROUP BY
        ap.seafarer_id,
        ap.contract_id,
        ap.created_at,
        ap.vessel_id,
        ap.to_date,
        ap.from_date,
        ap.template_id,
        ap.template_name,
        ap.status,
        sf.crew_code
)

WHERE performance_rating <> 0

GROUP BY
    seafarer_id,
    created_at,
    crew_code,
    to_date,
    from_date;
```

```python
%sql

INSERT OVERWRITE  TABLE reporting_layer.sac_prod_seafarer_public.digital_appraisal_view 

(
  SELECT
    DAV.*,
    PR.performance_rating
  FROM
    (
      SELECT
        *
      FROM
        (
          SELECT
            T1.*,
            ROW_NUMBER() OVER (PARTITION BY SEAFARER_ID ORDER BY FROM_DATE DESC) AS RNUM
          FROM
            (
              SELECT
                * EXCEPT (
                  `Crewing Superintendent Feedback`,
                  `Marine Superintendent Feedback`,
                  `Technical Superintendent Feedback`,
                  `Appraisee feedback`,
                  `Marine Manager Feedback`,
                  `Technical Manager Feedback`,
                  `Crewing Superintendent Feedback Status`,
                  `Marine Superintendent Feedback Status`,
                  `Technical Superintendent Feedback Status`,
                  `Appraisee feedback Status`,
                  `Marine Manager Feedback Status`,
                  `Technical Manager Feedback Status`
                ),

                CASE
                  WHEN
                    `Crewing Superintendent Feedback` IS NULL
                  THEN
                    MAX(`Crewing Superintendent Feedback`) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE `Crewing Superintendent Feedback`
                END AS `Crewing Superintendent Feedback`,

                CASE
                  WHEN
                    `Crewing Superintendent Feedback Status` IS NULL
                  THEN
                    MAX(split(`Crewing Superintendent Feedback Status`,'-')[0]) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE split(`Crewing Superintendent Feedback Status`,'-')[0]
                END AS `Crewing Superintendent Feedback Status`,

                CASE
                  WHEN
                    `Crewing Superintendent Feedback Status` IS NULL
                  THEN
                    MAX(split(`Crewing Superintendent Feedback Status`,'-')[1]) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE split(`Crewing Superintendent Feedback Status`,'-')[1]
                END AS `Crewing Superintendent Feedback Comment`,

                CASE
                  WHEN
                    `Marine Superintendent Feedback` IS NULL
                  THEN
                    MAX(`Marine Superintendent Feedback`) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE `Marine Superintendent Feedback`
                END AS `Marine Superintendent Feedback`,
                CASE
                  WHEN
                    `Marine Superintendent Feedback Status` IS NULL
                  THEN
                    MAX(split(`Marine Superintendent Feedback Status`,'-')[0]) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE split(`Marine Superintendent Feedback Status`,'-')[0]
                END AS `Marine Superintendent Feedback Status`,

                CASE
                  WHEN
                    `Marine Superintendent Feedback Status` IS NULL
                  THEN
                    MAX(split(`Marine Superintendent Feedback Status`,'-')[1]) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE split(`Marine Superintendent Feedback Status`,'-')[1]
                END AS `Marine Superintendent Feedback Comment`,
                
                CASE
                  WHEN
                    `Technical Superintendent Feedback` IS NULL
                  THEN
                    MAX(`Technical Superintendent Feedback`) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE `Technical Superintendent Feedback`
                END AS `Technical Superintendent Feedback`,
                CASE
                  WHEN
                    `Technical Superintendent Feedback Status` IS NULL
                  THEN
                    MAX(split(`Technical Superintendent Feedback Status`,'-')[0]) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE split(`Technical Superintendent Feedback Status`,'-')[0]
                END AS `Technical Superintendent Feedback Status`,

                CASE
                  WHEN
                    `Technical Superintendent Feedback Status` IS NULL
                  THEN
                    MAX(split(`Technical Superintendent Feedback Status`,'-')[1]) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE split(`Technical Superintendent Feedback Status`,'-')[1]
                END AS `Technical Superintendent Feedback Comment`,

                CASE
                  WHEN
                    `Appraisee feedback` IS NULL
                  THEN
                    MAX(`Appraisee feedback`) OVER (PARTITION BY SEAFARER_ID ORDER BY FROM_DATE DESC)
                  ELSE `Appraisee feedback`
                END AS `Appraisee feedback`,
                CASE
                  WHEN
                    `Appraisee feedback Status` IS NULL
                  THEN
                    MAX(split(`Appraisee feedback Status`,'-')[0]) OVER (PARTITION BY SEAFARER_ID ORDER BY FROM_DATE DESC)
                  ELSE split(`Appraisee feedback Status`,'-')[0]
                END AS `Appraisee feedback Status`,

                CASE
                  WHEN
                    `Appraisee feedback Status` IS NULL
                  THEN
                    MAX(split(`Appraisee feedback Status`,'-')[1]) OVER (PARTITION BY SEAFARER_ID ORDER BY FROM_DATE DESC)
                  ELSE split(`Appraisee feedback Status`,'-')[1]
                END AS `Appraisee feedback Comment`,

                CASE
                  WHEN
                    `Marine Manager Feedback` IS NULL
                  THEN
                    MAX(`Marine Manager Feedback`) OVER (PARTITION BY SEAFARER_ID ORDER BY FROM_DATE DESC)
                  ELSE `Marine Manager Feedback`
                END AS `Marine Manager Feedback`,
                CASE
                  WHEN
                    `Marine Manager Feedback Status` IS NULL
                  THEN
                    MAX(split(`Marine Manager Feedback Status`,'-')[0]) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE split(`Marine Manager Feedback Status`,'-')[0]
                END AS `Marine Manager Feedback Status`,

              CASE
                  WHEN
                    `Marine Manager Feedback Status` IS NULL
                  THEN
                    MAX(split(`Marine Manager Feedback Status`,'-')[1]) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE split(`Marine Manager Feedback Status`,'-')[1]
                END AS `Marine Manager Feedback comment`,

                CASE
                  WHEN
                    `Technical Manager Feedback` IS NULL
                  THEN
                    MAX(`Technical Manager Feedback`) OVER (PARTITION BY SEAFARER_ID ORDER BY FROM_DATE DESC)
                  ELSE `Technical Manager Feedback`
                END AS `Technical Manager Feedback`,
                CASE
                  WHEN
                    `Technical Manager Feedback Status` IS NULL
                  THEN
                    MAX(split(`Technical Manager Feedback Status`,'-')[0]) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE split(`Technical Manager Feedback Status`,'-')[0]
                END AS `Technical Manager Feedback Status`,

              CASE
                  WHEN
                    `Technical Manager Feedback Status` IS NULL
                  THEN
                    MAX(split(`Technical Manager Feedback Status`,'-')[1]) OVER (
                        PARTITION BY SEAFARER_ID
                        ORDER BY FROM_DATE DESC
                      )
                  ELSE split(`Technical Manager Feedback Status`,'-')[1]
                END AS `Technical Manager Feedback Comment`

              FROM
                (
                  SELECT
                    *
                  FROM
                    reporting_layer.sac_prod_seafarer_public.add_digital_appraisal_view
                      PIVOT (
                        MAX(concat(`feedback Status`, '-', FEEDBACK_COMMENTS)) FOR `feedback_templateName` IN (
                          'Crewing Superintendent Feedback' AS `Crewing Superintendent Feedback Status`,
                          'Marine Superintendent Feedback' AS `Marine Superintendent Feedback Status`,
                          'Technical Superintendent Feedback' AS `Technical Superintendent Feedback Status`,
                          'Appraisee feedback' AS `Appraisee feedback Status`,
                          'Marine Manager Feedback' AS `Marine Manager Feedback Status`,
                          'Technical Manager Feedback' AS `Technical Manager Feedback Status`
                        )
                      )
                ) T
            ) T1
        )
      WHERE
        RNUM = 1
    ) DAV
    LEFT JOIN (
      SELECT 
        seafarer_id, 
        ROUND(AVG(performance_rating), 2) AS performance_rating
      FROM reporting_layer.sac_prod_seafarer_public.appraisal_performance
      GROUP BY seafarer_id
    ) PR
      ON DAV.SEAFARER_ID = PR.seafarer_id
)
```
